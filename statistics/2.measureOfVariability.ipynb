## Mean absolute Deviation

#### the mean absolute deviation of a dataset is the averge distance between each data point and the mean.it  gives us an idea about the variability in a dataset.

```python
sec_a=np.array([75,65,73,68,72,67])
sec_b=np.array([90,47,43,96,93,51])
```

```python
mean=np.mean(sec_b)
```

```python
no=np.array([1,2,3,4,5,6])

plt.figure(figsize=(10,3))
plt.scatter(sec_a,no,color='blue',label='Section A')
plt.scatter(sec_b,no,color='red',label='Section B')
plt.axvline(mean, linestyle='--',label="Mean")
plt.legend()
```

```python
mad_a=np.sum(np.abs(sec_a-mean))/len(sec_a)
mad_b=np.sum(np.abs(sec_b-mean))/len(sec_b)
```

```python
mad_a,mad_b
```

#### mad_b -so we use this as this has low skwness

*when there is situation where mad_a == mad_b then we use standard deviation in such situation*

# standard deviation and variance


*square of std is variance ..so varience help us to find who has bigger skwness*

```python
np.std(sec_a),np.std(sec_b)
```

```python
np.var(sec_a),np.var(sec_b)
```
