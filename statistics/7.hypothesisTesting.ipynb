### Hypothesis testing Z-test
*the teacher clm that the mean score of the student is greater then 82 adn standard deviation is 20.if sample of 81 stuent was taken with the mean score of 92..*

```python
import scipy.stats as st
import numpy as np
```

```python
s_x=90
p_u=82
p_std=20
n=81
ap=0.05
```

```python
z_cal=(s_x-p_u)/(p_std/np.sqrt(n))
```

```python
z_table=st.norm.ppf(1-ap)
z_table
```

```python
if z_table <z_cal:
    print("H1 is right")
else:
    print("H0 is right")    
```

## Chi square test


*Why to use*
1. Goodness
2. Independence


Hypothesis testing CHi square tes


exampe 1:
    A fair dice is rolled 120 ties and the following results are obtained:
        Face 1:22 times
        Face 2:17 times
        Face 3:20 times
        face 4:26 times
        face 5:22 times
        face 6:13 times
        
test at a 5% level of significance wether the dice is fair

H0 is fair
H1 is unfair

*df=n-1*

```python
ob=np.array([22,17,20,26,22,13])
ex=np.array([20,20,20,20,20,20])
```

```python
chi_test=np.sum(np.square(ob-ex)/ex)
chi_test
```

```python
chi_critical = st.chi2.ppf(1 - 0.05, df=5)
chi_critical
```

```python
if chi_test > chi_critical:
    print("Reject H0")
    print("Conclusion: The dice is NOT fair.")
else:
    print("Fail to reject H0")
    print("Conclusion: The dice CAN be considered fair.")
```
