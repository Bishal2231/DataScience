# MEAN

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ar=np.array([7000,7500,2000,3000,4000,4000,8000])
```

```python
ar
```

```python
mean=np.mean(ar)
mean
```

```python
df=pd.read_csv('train.csv')
df.head()
```

```python
df['Age'].mean()
```

```python
mn=np.mean(df["Age"])
```

```python
sns.histplot(x="Age",data=df,bins=10)
plt.axvline(mn, linestyle='--')
```

# Median


```python
np.median(ar) 
```

```python
df["Age"].isna().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
```

```python
md=np.median(df['Age'])
```

```python
sns.histplot(x="Age",data=df,bins=10)
plt.axvline(mn,color="red",linestyle='--',label="Mean")
plt.axvline(md,color="blue",linestyle='--',label="Median")
plt.legend()
```
