# Skewness

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('train.csv')
```

```python
sns.histplot(x="Age",data=df)
```

```python
df['Age'].skew()

# if value is greather then 0,then its positive skewed
#if value is less then 0,then its negative skewed
# if value is 0,then its normal distribution
```

```python
data=np.random.normal(0,100,100)
```

```python
data
```

```python
df=pd.DataFrame({"x":data})
df.skew()
```

```python
sns.histplot(x="x",data=df)
```
